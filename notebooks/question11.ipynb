```python
import numpy as np

def compute_mean(values):
    mean_value = np.mean(values)
    print(mean_value)

list_mean = compute_mean([2, 4, 6, 8])
```

## How this function works
### The function
The `compute_mean` function works by calculating the mean (using the function from the numpy library) based on a list of values provided. The resulting mean value is then printed below the cell.
### The equation
The equation for calculation of the mean is expressed in two ways below.

$$
\bar{x} = \frac{x_1 + x_2 + ... + x_n}{n} = \frac{1}{n} \left(\sum_{i=1}^{n} x_i \right)
$$
